## Array Slicing

```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6, 7])

print(arr[1:5])
print(arr[4:])
print(arr[:4])
print(arr[-3:-1])
print(arr[1:5:2])
print(arr[::2])
print(arr[0:2])
```

## NumPy Array Copy vs View
The main difference between a copy and a view of an array is that the copy is a new array, and the
view is just a view of the original array.

The copy SHOULD NOT be affected by the changes made to the original array.

The view SHOULD be affected by the changes made to the original array.

```python
# copy()
arr = np.array([1, 2, 4, 4, 5])
x = arr.copy()

arr[2] = 3

print(arr)
print(x)
```

```python
# view()
arr = np.array([1, 2, 4, 4, 5])
x = arr.view()

arr[2] = 3

print(arr)
print(x)
```

```python
# view()
arr = np.array([1, 2, 4, 4, 5])
x = arr.view()

x[2] = 3

print(arr)
print(x)
```

## Joining NumPy Arrays

```python
import numpy as np

arr1 = np.array([1, 2, 3, 4])
arr2 = np.array([5, 6, 7, 8])
arr3 = np.concatenate((arr1, arr2)) # Join two arrays
print("Concatenate:\n", arr3)

arr4 = np.stack((arr1, arr2), axis=1) # Joining Arrays Using Stack Functions
print("\nStack:\n", arr4)

arr5 = np.hstack((arr1, arr2)) # NumPy provides a helper function: hstack() to stack along rows.
print("\nH-Stack:\n", arr5)

arr6 = np.vstack((arr1, arr2)) # NumPy provides a helper function: vstack() to stack along columns.
print("\nV-Stack:\n", arr6)

arr7 = np.dstack((arr1, arr2)) # NumPy provides a helper function: dstack() to stack along height, which is the same as depth.
print("\nD-Stack:\n", arr7)
```

## Searching Arrays

```python
arr = np.array([1, 2, 3, 4, 5, 6])

x1 = np.where(arr == 4) # To search an array, use the where() method
print("Array is 4 at index:", x1)

x2 = np.where(arr % 2 == 0) # Find the indexes where the values are even
print("Array modulas 0 is:", x2)

x3 = np.where(arr % 2 == 1) # Find the indexes where the values are odd.
print("Array modulas 1 is:", x3)

x4 = np.searchsorted(arr, 6) # The searchsorted() method is assumed to be used on sorted arrays. Find the indexes where the value 6 should be inserted.
print(x3)
```
