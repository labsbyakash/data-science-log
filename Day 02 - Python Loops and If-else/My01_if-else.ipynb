# Python If - Else

```python
a = 45
b = 100
if b > a:
  print("b is greater than a")
```

```python
a = 24
b = 24
if b > a:
  print("b is greater than a")
elif a == b:
  print("a and b are equal")
```

```python
a = 100
b = 45
if b > a:
  print("b is greater than a")
elif a == b:
  print("a and b are equal")
else:
  print("a is greater than b")
```
