# Python Loop

Python has two primitive loop commands: while loops for loops

---
## while Loop

```python
x = 1
while x < 5:
    print(x)
    x = x + 1
```

```python
# With the break statement we can stop the loop even if the while condition is true.
x = 1
while x < 5:
    print(x)
    x = x + 1
    if (x == 3):
        break
```

```python
# With the continue statement we can stop the current iteration, and continue with the next.
x = 0
while x < 5:
    x = x + 1
    if (x == 3):
        continue
    print(x)
```

```python
guess_the_num = random.randrange(1, 11)

guessed_num = 0

print("Guess the number between 1 and 10!")

while guessed_num != guess_the_num:
    guessed_num = int(input("Enter Number: "))
    if (guessed_num == guess_the_num):
        print("Congratulations Babe!, you've done well")
        break
    elif (guessed_num > guess_the_num):
        print("Your number is bigger")
    else:
        print("Your number is smaller")
```

## for Loop

Ways to write for loop

```python
x = [1, 2, 3, 4]

for i in x:
    print(i)
```

```python
for i in range(5):
    print(i)
```

```python
for i in range(1, 5):
    print(i)
```

```python
for i in range(1, 10, 2): # skip the every 2 step
    print(i)
```

```python
names = ["Light", "Madara", "L"]

for index, name in enumerate(names):
    print(index, name)

print("")

# if want to start with 1
for index, name in enumerate(names, start = 1):
    print(index, name)
```

```python
dictt = {
    "name": "Akash",
    "age": 24,
    "has_done": False
}

# keys only
for key in dictt:
    print(key)

print()

# keys with value
for key, value in dictt.items():
    print(f"{key} : {value}")
```

```python
# print 2 list
names = ["Akash", "Rahul"]
ages = [24, 22]

for name, age in zip(names, ages):
    print(name, age)
# Akash 24
# Rahul 22
```

```python
square = []
for i in range(1, 5):
    square.append(i * i)

print(square)

# one liner
new_sqre = [i * i for i in range(1, 5)]
print(new_sqre)
```

## Print star '*' using for loop

```python
# triangle star
def print_tri_star(n):
    k = 2 * n - 2
    for i in range(0, n):
        for _ in range(0, k):
            print(end=" ")
        k = k - 1
        for _ in range(0, i + 1):
            print("*", end=" ")
        print(" ")

print_tri_star(10)
```

```python
# better approach
def triangle_shape(n):
    for i in range(n):
        spaces = " " * (n - i - 1)
        stars = "* " * (i + 1)
        print(spaces, stars)

triangle_shape(10)
```

```python
# right triangle up + down
def right_triangle_shape(n):
    # Upar wala - badhta hua
    for i in range(1, n + 1):
        print("* " * i)

    # Neeche wala - ghatta hua
    for i in range(n - 1, 0, -1):
        print("* " * i)

right_triangle_shape(5)
```

```python
# Sandglass
def sandglass_shape(n):
    for i in range(n, 0, -1):
        spaces = " " * (n - i)
        stars = "* " * i
        print(spaces, stars)

    for i in range(2, n + 1):
        spaces = " " * (n - i)
        stars = "* " * i
        print(spaces, stars)

sandglass_shape(5
)
```

```python
# pant style star

def pant_shape(n):
    for i in range(n):
        left = "* " * (n - i)
        middle = "  " * (2 * i)
        right = "* " * (n - i)
        print(left + middle + right)

    for i in range(n - 1, -1, -1):
        left = "* " * (n - i)
        middle = "  " * (2 * i)
        right = "* " * (n - i)
        print(left + middle + right)

pant_shape(10)
```
