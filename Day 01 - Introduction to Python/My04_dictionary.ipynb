# Python Dictionary

Dictionary is used to store data values in key:value pairs. A dictionary is a collection which is ordered, changeable and do not allow duplicates. Dictionaries are written with curly brackets, and have keys and values.

```python
dictt = {
    "brand" : "Ford",
    "Model" : "Mustang",
    "Year" : 1964,
    "Colors" : ["red", "blue", "black"]
}

print(dictt)
print("Length: ", len(dictt))
print("Type", type(dictt))
```

---
## Access Dictionary Items

You can access the items of a dictionary by referring to its key name, inside square brackets.

```python
dictt = {
    "brand" : "Ford",
    "Model" : "Mustang",
    "Year" : 1964,
    "Colors" : ["red", "blue", "black"]
}

print(dictt["brand"])
print(dictt.get("brand"))
print(dictt.keys())
print(dictt.values())
print(dictt.items())
```

---
## Change Dictionary Items

You can change the value of a specific item by referring to its key name.

```python
dictt = {
    "brand" : "Ford",
    "Model" : "Mustang",
    "Year" : 1964,
    "Colors" : ["red", "blue", "black"]
}

dictt['Year'] = 2026

print(dictt)
```

---
## Adding Items

Adding an item to the dictionary is done by using a new index key and assigning a value to it.

```python
dictt["is_bestseller"] = True
print(dictt)
```

```python
dictt['year'] = 1969
print(dictt)

dictt['Year'] = 1969
print(dictt)
```

---
## Remove Dictionary Items

The pop() method removes the item with the specified key name: The popitem() method removes the last inserted item (in versions before 3.7, a random item is removed instead): The del keyword removes the item with the specified key name: The del keyword can also delete the dictionary completely: The clear() method empties the dictionary.

```python
dictt.pop("year")
print(dictt)
```

```python
dictt['year'] = 1969
print(dictt)

dictt.popitem()
print(dictt)
```

```python
dictt['year'] = 1969
print(dictt)

del dictt['year']
print(dictt)
```

```python
dictt.clear()
print(dictt)
```
