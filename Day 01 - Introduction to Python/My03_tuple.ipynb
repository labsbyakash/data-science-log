# Python Tuple

Tuple items are ordered, unchangeable, and allow duplicate values.

Tuples are written with round brackets.

```python
py_tup = (1, 2, "variables")
print(py_tup)
print("Length:", len(py_tup))
print("Type:", type(py_tup))

# Note: The values True and 1 are considered the same value in sets, and are treated as duplicates:
py_tup1 = (1, 2, 3, True, "Four", "False", 20.20)
print(py_tup1)
print("Type:", type(py_tup1))
print("Length:", len(py_tup1))
```

```python
# We can also create tuple using tuple() constructor along with double brackets.
tup_cons = tuple((1, 2, 3, 4, 5))
print(type(tup_cons))
print(tup_cons)
```

---
## Access Tuple Items

```python
fruits = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")
print(fruits[1])
print(fruits[-1])
print(fruits[2:5])
print(fruits[:4])
print(fruits[2:])
print(fruits[-4:-1])
```

---
## Update Tuples

Once a tuple is created, we cannot change its values. Tuples are unchangeable, or immutable as it also is called.

But there is a trick we can do is that, we'll convert the tuple into a list, change the list, and convert the list back into a tuple.

```python
tup_var = (1, 2, 3, 4)
li = list(tup_var)
li.append(5)
tup_var = tuple(li)
print(tup_var)
```

```python
tup_var = (1, 2, 3, 4)
li = list(tup_var)
li.remove(4)
tup_var = tuple(li)
print(tup_var)
```

## Unpack Tuples

When we create a tuple, we normally assign values to it. This is called "packing" a tuple: But, in Python, we are also allowed to extract the values back into variables. This is called "unpacking":

```python
students = [("Akash", 24, "A+"), ("Rahul", 22, "B+")]

for name, _, grade in students:
    print(name, grade)

# Output:
# Akash A+
# Rahul B+
```

```python
#Using Asterisk*
#Assign the rest of the values as a list called "red":

fruits = ("apple", "banana", "cherry", "strawberry", "raspberry")

(green, yellow, *red) = fruits

print(green)
print(yellow)
print(red)
```

## Loop Tuples

Print all items by referring to their index number: Print all items, using a while loop to go through all the index numbers:

```python
tup = ("Light", "L", "Madara")
for i in range(len(tup)):
    print(tup[i])

print("\nNow print with _\n")

for _ in range(len(tup)):
    print(tup[_])
```

```python
tup = ("Light", "L", "Madara")
i = 0;

while i < len(tup):
    print(tup[i])
    i = i + 1
```
