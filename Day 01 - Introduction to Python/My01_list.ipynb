# Python Lists

Python list is a built-in data structure used to store an ordered, mutable collection of items inside square brackets. Lists allow duplicate values and can hold items of any data type—including numbers, strings, booleans, and even other lists—all within the same container.

### Print List

```python
li = ['akash', 'anmol', 'shinobi', 'lord']
print(li)

#types
print(type(li))

# print specific element
print(li[0])
print(li[1])
print("\nPrint end:")
print(li[-1])
print("\nPrint from 2 to end")
print(li[1:])

# check if it's having that element in the list
if "akash" in li:
    print("Yes, it's having the text word 'akash'")
else:
    print("No, it's not having the text word 'akash'")
```

### Add List Items

append() insert() extend()

```python
# append
li.append('madara')
print(li)
```

```python
# insert
li.insert(3, 'light')
print(li)
```

```python
# extend
li2 = ['kun', 'san', 'chan']
li2.extend(li)
print(li2)
```

```python
# Add elements of a tuple to a list:
li_tup = (2, 4, "avengers")
li2.extend(li_tup)
print(li2)
```

### Remove List Items

remove(): it removes the specified item. <br />
pop(): it removes the specified index. <br />
del: this keyword also removes the specified index:(can also delete the list completely.) <br />
clear(): this method empties the list (The list still remains, but it has no content)

```python
#Remove the first occurance
li.append('madara')
print(li)

li.remove('madara')
print(li)
```

```python
# Remove Specified Index
#The pop() method removes the specified index.
li.pop(5)
print(li)

```

```python
del li[1]
print(li)
```

```python
# clear() method empties the list.
new_li = [1, 2, 3, 4]
print(new_li)

new_li.clear()
print(new_li)
print(type(new_li))
```

### Sort Lists

sort(): To sort descending, use the keyword argument reverse = True <br />
reverse(): this method reverses the current sorting order of the elements.

```python
print("Before sort()")
print(li)

print("After sort()")
li.sort()
print(li)
```

```python
# sort(reverse=True)
print("Before sort(reverse=True)")
print(li)

print("After sort(reverse=True)")
li.sort(reverse=True)
print(li)
```

```python
# reverse
print("Before reverse()")
print(li)

print("After reverse()")
li.reverse()
print(li)
```

### Join Two Lists

using +, append() and extend()

```python
list1 = ['a', 'b', 'c']
list2 = [1, 2, 3]

list3 = list1 + list2
print(list3)
```

```python
for i in list2:
    list1.append(i)

print(list1)
```

```python
list1 = ["a", "b" , "c"]
list2 = [1, 2, 3]

list1.extend(list2)
print(list1)
```

### Copy Lists

using copy() methond and list() function

```python
list1_copy = list1.copy()
print(list1_copy)
```

```python
sett = (1, 2, 3, 4, 4)
list1_copy = list(sett)
print(list1_copy)
```
