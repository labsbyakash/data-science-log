# Python Sets

A set is a collection which is unordered, unchangeable*, and Duplicates Not Allowed. Sets are used to store multiple items in a single variable.

Note: Set items are unchangeable, but we can remove items and add new items. Sets are written with curly brackets.

```python
py_sets = {1, 2, "variables"}
print(py_sets)
print("Length:", len(py_sets))
print("Type:", type(py_sets))

# Note: The values True and 1 are considered the same value in sets, and are treated as duplicates:
py_sets1 = {1, 2, 3, True, "Four", "False", 20.20}
print(py_sets1)
print("Type:", type(py_sets1))
print("Length:", len(py_sets1))
```

```python
# Using the set() constructor to make a set
set2 = set(("Akash", "Sharma", 24))
print(set2)
```

---
## Access Set Items

We cannot access items in a set by referring to an index or a key.

But We can loop through the set items using a for loop, or ask if a specified value is present in a set, by using the in keyword.

```python
print("Akash" in set2)
```

---
## Add Set Items

Once a set is created, you cannot change its items, but you can add new items.

- Add an item to a set, using the add() method:
- To add items from another set into the current set, use the update() method.

```python
py_sets.add("Akashi")
print(py_sets)
```

```python
fruits = {"apple", "banana", "cherry"}
fruits2 = {"pineapple", "mango", "papaya"}
fruits.update(fruits2)
print(fruits)
```

---
## Remove Set Items

- Remove "L" by using the remove() method.
- Remove "L" by using the discard() method.
- Remove a random item by using the pop() method.
- The clear() method empties the set.
- The del keyword will delete the set completely.

```python
ani_char = {"Light", "Madara", "L"}
ani_char.remove("L")
print(ani_char)
```

```python
ani_char = {"Light", "Madara", "L"}
ani_char.discard("L")
print(ani_char)
```

```python
ani_char = {"Light", "Madara", "L"}
valuee = ani_char.pop()
print(valuee)
print(ani_char)
```

```python
ani_char = {"Light", "Madara", "L"}
ani_char.clear()
print(ani_char)
```

```python
ani_char = {"Light", "Madara", "L"}
del ani_char # it'll delete the variable
```

## Join Two Sets

- The union() method returns a new set with all items from both sets.
- The update() method inserts the items in set2 into set1: (Keep ONLY the Duplicates).
- The intersection_update() method will keep only the items that are present in both sets. (Keep All, But NOT the Duplicates).
- The symmetric_difference_update() method will keep only the elements that are NOT present in both sets.

```python
set1 = {"a", "b" , "c"}
set2 = {1, 2, 3}
set3 = set1.union(set2)
print(set3)
```

```python
set1 = {1, 2, 3, 4}
set2 = {4, 5, 6, 7}
set1.intersection_update(set2)
print(set1)
```

```python
set1 = {1, 2, 3, 4}
set2 = {4, 5, 6, 7}
set1.symmetric_difference_update(set2)
print(set1)
```

```python
set1 = {"a", "b" , "c"}
set2 = {1, 2, 3}
set1.update(set2)
print(set1)
```
