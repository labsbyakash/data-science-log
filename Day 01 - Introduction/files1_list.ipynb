# Python Lists

Python list is a built-in data structure used to store an ordered, mutable collection of items inside square brackets. Lists allow duplicate values and can hold items of any data type—including numbers, strings, booleans, and even other lists—all within the same container.

### Print List

```python
li = ['akash', 'anmol', 'shinobi', 'lord']
print(li)

#types
print(type(li))

# print specific element
print(li[0])
print(li[1])
print("\nPrint end:")
print(li[-1])
print("\nPrint from 2 to end")
print(li[1:])

# check if it's having that element in the list
if "akash" in li:
    print("Yes, it's having the text word 'akash'")
else:
    print("No, it's not having the text word 'akash'")
```

### Add List Items

append() insert() extend()

```python
# append
li.append('madara')
print(li)
```

```python
# insert
li.insert(3, 'light')
print(li)
```

```python
# extend
li2 = ['kun', 'san', 'chan']
li2.extend(li)
print(li2)
```

```python
# Add elements of a tuple to a list:
li_tup = (2, 4, "avengers")
li2.extend(li_tup)
print(li2)
```

### Remove List Items

remove(): it removes the specified item. <br />
pop(): it removes the specified index. <br />
del: this keyword also removes the specified index:(can also delete the list completely.) <br />
clear(): this method empties the list (The list still remains, but it has no content)

```python
#Remove the first occurance
li.append('madara')
print(li)

li.remove('madara')
print(li)
```

```python
# Remove Specified Index
#The pop() method removes the specified index.
li.pop(5)
print(li)

```
