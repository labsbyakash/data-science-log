## Creating a Class
A class is a blueprint or a template for creating objects while an object is an instance or a copy of the class with actual values.

```python
# Create a class using the class keyword.
class Details:
    name = "Aryan"
    address = "Delhi"
    age = 25
```

---
## Creating an Object:
create object of the class.

```python
class Details:
    name = "Aryan"
    address = "Delhi"
    age = 25

obj1 = Details()
print(obj1.name)
print(obj1.address)
print(obj1.age)
```

---
## self method
The self parameter is a reference to the current instance of the class, and is used to access variables that belong to the class.

It must be provided as the extra parameter inside the method definition.

```python
class Details:
    name = "Aryan"
    address = "Delhi"
    age = 25

    def desc(self):
        print("My name is", self.name,
              "I'm from",self.address,
              "and I'm", self.age, "years old.")

obj1 = Details()
obj1.desc()
```

---
## init method
The init method is used to initialize the object’s state and contains statements that are executed at the time of object creation.

```python
# The string representation of an object WITHOUT the __str__() function:
class Details:
    def __init__(self, name, age):
        self.name = name
        self.age = age

obj1 = Details("John", 36)
print("Hello my name is " + obj1.name)
```

```python
# The string representation of an object WITH the __str__() function:
class Details:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __str__(self):
        return f"{self.name}({self.age})"

obj1 = Details("John", 36)

print(obj1)
```

## Python Iterators
Iterators in python are used to iterate over iterable objects or container datatypes like lists, tuples, dictionaries, sets, etc.

It consists of iter() and next() methods.

iter() : to initialize an iterator, use the iter() method.

next() : This method returns the next item of the sequence.



```python
# Using inbuilt iterators:
string = 'Hello World'
iterObj = iter(string)

while True:
    try:
        char1 = next(iterObj)
        print(char1)
    except StopIteration:
        break
```

## Python Try Except
The try block lets you test a block of code for errors.

The except block lets you handle the error.

The else block lets you execute code when there is no error.

The finally block lets you execute code, regardless of the result of the try- and except blocks.

```python
try:
    num = int(input("Enter an integer: "))
except ValueError:
    print("Number entered is not an integer.")
```

```python
try:
    num = int(input("Enter an integer: "))
except ValueError:
    print("Number entered is not an integer.")
else:
    print("Integer Accepted.")
```

```python
try:
    num = int(input("Enter an integer: "))
except ValueError:
    print("Number entered is not an integer.")
else:
    print("Integer Accepted.")
finally:
    print("This block is always executed.")
```
