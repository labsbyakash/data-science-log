## Creating a Class

A class is a blueprint or a template for creating objects while an object is an instance or a copy of the class with actual values.

```python
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

my_car = Car("Toyota", "Corolla")
print(f"My car is a {my_car.brand} {my_car.model}.")
```

```python
class Car:
    def __init__(self, brand, model):
        self.brand = brand
        self.model = model

    def display_info(self):
        return f"My car is a {self.brand} {self.model}."

my_car = Car("Toyota", "Corolla")
print(my_car.display_info())
```

```python
# inheritance example

class ElectricCar(Car):
    def __init__ (self, brand, model, battery_size):
        super().__init__(brand, model)
        self.battery_size = battery_size

my_electric_car = ElectricCar("Tesla", "Model S", 100)
print(f"My electric car is a {my_electric_car.brand} {my_electric_car.model} with a {my_electric_car.battery_size}-kWh battery.")
print(my_electric_car.display_info())
```

```python
# encapsulation example
class Car:
    def __init__(self, brand, model):
        self.__brand = brand
        self.model = model

    def get_brand(self):
        return self.__brand + " (This is a private attribute.)"

    def display_info(self):
        return f"My car is a {self.__brand} {self.model}."

class ElectricCar(Car):
    def __init__ (self, brand, model, battery_size):
        super().__init__(brand, model)
        self.battery_size = battery_size

my_electric_car = ElectricCar("Tesla", "Model S", 100)
# print(f"My electric car is a {my_electric_car.brand} {my_electric_car.model} with a {my_electric_car.battery_size}-kWh battery.")
# print(my_electric_car.display_info())
print(my_electric_car.get_brand())
```

```python
# polymorphism example - Same function name but different implementation in different classes
class Car:
    def __init__(self, brand, model):
        self.__brand = brand
        self.model = model

    def get_brand(self):
        return self.__brand + " (This is a private attribute.)"

    def display_info(self):
        return f"My car is a {self.__brand} {self.model}."

    def fuel_type(self):
        return "This car uses petrol or diesel."

class ElectricCar(Car):
    def __init__ (self, brand, model, battery_size):
        super().__init__(brand, model)
        self.battery_size = battery_size

    def fuel_type(self):
        return "This car is electric."

my_electric_car = ElectricCar("Tesla", "Model S", 100)
print(my_electric_car.fuel_type())

new_car = Car("Toyota", "Corolla")
print(new_car.fuel_type())
```

```python
# variables in oops

class Car:
    car_counter = 0  # class variable to count the number of car instances
    
    def __init__(self, brand, model):
        self.__brand = brand
        self.model = model
        Car.car_counter += 1

    def get_brand(self):
        return self.__brand + " (This is a private attribute.)"

    def display_info(self):
        return f"My car is a {self.__brand} {self.model}."

    def fuel_type(self):
        return "This car uses petrol or diesel."

class ElectricCar(Car):
    def __init__ (self, brand, model, battery_size):
        super().__init__(brand, model)
        self.battery_size = battery_size

    def fuel_type(self):
        return "This car is electric."

# my_electric_car = ElectricCar("Tesla", "Model S", 100)
# print(my_electric_car.fuel_type())

Car("Toyota", "Corolla")
# print(new_car.fuel_type())

print(f"Total number of cars created: {Car.car_counter}")
```
